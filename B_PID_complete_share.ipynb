```python
# check that this is one Sample Number for one Patient ID...
duplicate_Sample_Number_PID = df_PID['Sample Number'].duplicated()
print(duplicate_Sample_Number_PID.value_counts())
# Nope! 105 instances of duplicate Sample Numbers
# I think this can only happen if the same Sample Number has been registered to two pateint records...
# ... possibly through merge in Hematos
```

```python
duplicate_patientID_PID = df_PID['Patient ID'].duplicated()
print(duplicate_patientID_PID.value_counts())
# Nope! 369 instances of a duplicate on Patient ID
# Suggestive 369 patients have been repeat tested on new Sample Numbers
```

```python
duplicate_NHS_Number_PID = df_PID['NHS Number'].duplicated()
print(duplicate_NHS_Number_PID.value_counts())
# 470 duplicates of NHS number, probably because pateints were re-tested on new Sample numbers
```

```python
df_PID[df_PID.duplicated(subset=['Sample Number', 'Patient ID', 'NHS Number'], keep=False) == True].sort_values('Sample Number')
# zero duplicates exist across all 3 criteria
```

```python
df_PID[df_PID.duplicated(subset=['Sample Number', 'Patient ID'], keep=False) == True].sort_values('Sample Number')
# zero duplicates exist across those 2.  This makes sense as a genotype record can only be created a Sample Number
# and a sample number should be unique to a Pateint ID (see below...)
```

```python
df_PID[df_PID.duplicated(subset=['Patient ID', 'NHS Number'], keep=False) == True].sort_values('Patient ID')
# 706 rows
# These duplicates simply represent the same pateint being re-tested on a different sample Number
# It must always be a different Sample Number as the search for all 3 categories returned zero
# this is cool
# when df_PID is merged with df_hgp the sample with the least complete genotype will not merge
# as the matching Sample Number has been deleted from df_hgp
# or the two Sample Numbers will align in both datasets, with identical PID and hgpdata
# NOTE TO SELF
# These duplicates will have to be removed once the datasets are merged/combined
```

```python
df_PID[df_PID.duplicated(subset=['Sample Number', 'NHS Number'], keep=False) == True].sort_values('Sample Number')
# 201 across those 2
# These duplicates are caused by merging two Hematos records (each with a unique Pateint ID) 
# the PID data will be merged to df_hgp based upon Sample Number
# it is important to have no duplicates in df_PID at Sample Number or the merge will not work
# sense check these duplicates 
```
